```python
import os
import matplotlib.pyplot as plt
import torch
from transformers import DistilBertTokenizer, DistilBertModel
```

```python
sentences = [
    "She is a bad girl.",
    "Krishnochurar bonyay Chaitali bhese geche."
]
```

```python
device = torch.device('cuda')
```

```python
device
```

```python
model=DistilBertModel.from_pretrained("distilbert-base-uncased")
model=model.to(device)
tokenizer=DistilBertTokenizer.from_pretrained("distilbert-base-uncased")
```

```python
model = model.eval()    
```

```python
inputs = tokenizer(sentences, return_tensors="pt", padding=True, truncation=True, max_length=50)
```

```python
inputs = inputs.to(device)
inputs
```

```python
sentences_1 = ['[CLS] ' + text + ' [SEP]' for _, text in enumerate(sentences)]
sentences_1
```

```python
tokenized_text = tokenizer.tokenize(sentences[0])
indexed_tokens = tokenizer.convert_tokens_to_ids(tokenized_text)
indexed_tokens
```

```python
with torch.no_grad():
    outputs = model(**inputs)

outputs
```

```python
embeddings = outputs.last_hidden_state[:, 0, :]
embeddings
```

```python
inputs_1 = tokenizer(sentences_1, return_tensors="pt", padding=True, truncation=True, max_length=50)
inputs_1 = inputs_1.to(device)
with torch.no_grad():
    outputs_1 = model(**inputs)

outputs_1
```

```python
embeddings_1 = outputs_1.last_hidden_state[:, 0, :]
embeddings_1
```

```python
outputs_1.last_hidden_state[0, :, :].size()
```
